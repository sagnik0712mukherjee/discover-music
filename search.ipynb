```python
import json
from sentence_transformers import SentenceTransformer

# Perform cosine similarity search
from sklearn.metrics.pairwise import cosine_similarity as cosine_similarity_fn
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
with open('data/JSON/all_songs_vectors.json', 'r') as f:
    all_songs_vectors = json.load(f)
print(len(all_songs_vectors))
```

```python
def write_results(text, file_name):
    with open(f'data/SEARCH_RESULTS/{file_name}', 'a') as f:
        f.write(text + '\n')
```

```python
query = "Romantic Ballad Song"
zone = "romantic_ballad_song"

filters = {
    "singers": None, 
    "year": None
}
encoding = model.encode([query])[0].tolist()

search_results = []
for song in all_songs_vectors:
    #check filter
    if (filters['singers'] and filters['singers'].lower() not in song['singers'].lower()) or (filters['year'] and filters['year'] != song['year']):
        continue
    song_vector = song['description_vector']
    similarity = cosine_similarity_fn([encoding], [song_vector])[0][0]
    search_results.append((song, similarity))

search_results.sort(key=lambda x: x[1], reverse=True)

for res in search_results[:10]:
    write_results(f"""
          Song: {res[0]['song_name']},
          Year Of Release: {res[0]['year']},
          Movie: {res[0]['movie']},
          YouTube Link: {res[0]['youtube_link']},
          Similarity: {res[1]}
          """, f'search_results_{zone}.txt')
```
